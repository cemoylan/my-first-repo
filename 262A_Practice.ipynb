```python
path = "/content/drive/MyDrive/Library_Usage.csv"
import pandas as pd
df = pd.read_csv(path)
df.head()
```

```python
df.info()
```

```python
df["Patron Type Definition"].value_counts()
```

```python
df["Total Checkouts"].value_counts()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="pie",
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Year Patron Registered"].value_counts().plot(
    kind="bar",
    color="pink",
    figsize=(11,8),
)
plt.title("Patron Registrations by Year")
plt.show()
```

```python
import matplotlib.pyplot as plt
df["Circulation Active Month"].value_counts().plot(
    kind="bar",
    color="orange",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Month")
plt.xlabel("Month")
plt.ylabel("# of Circulations")
plt.show()
```

```python
month_order=[
    "January","February","March","April","May","June","July","August","September","October","November","December"
    ]
df["Circulation Active Month"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="orange",
    edgecolor="pink",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Month")
plt.xlabel("Month")
plt.ylabel("# of Circulations")
plt.show()
```

```python
df.isna().sum()
```

```python
df["Circulation Active Month Cleaned"] = df["Circulation Active Month"].fillna('Not Captured')
df.isna().sum()
```

```python
month_order=[
    "January","February","March","April","May","June","July","August","September","October","November","December","Not Captured"
    ]
df["Circulation Active Month Cleaned"].value_counts().reindex(month_order).plot(
    kind="bar",
    color="yellow",
    edgecolor="darkblue",
    figsize=(11,8),
)
plt.title("Library Circulation Activity by Month")
plt.xlabel("Month")
plt.ylabel("# of Circulations")
plt.show()
```
